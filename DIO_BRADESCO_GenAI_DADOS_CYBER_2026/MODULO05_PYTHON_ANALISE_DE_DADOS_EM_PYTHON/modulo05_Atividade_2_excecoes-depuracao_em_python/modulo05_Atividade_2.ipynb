## O que são erros?

Durante a execução de um programa podem ocorrer **erros inesperados**.

Esses erros são chamados de **exceções**.

Eles acontecem quando o Python encontra uma situação que não consegue executar.

Exemplos:

- dividir um número por zero
- acessar um arquivo que não existe
- converter texto para número de forma incorreta

```python
# exemplo de erro

numero = 10 / 0
print(numero)
```

## Tipos de erros em Python

Alguns erros comuns são:

| Tipo de erro | Descrição |
|---|---|
| SyntaxError | erro de sintaxe |
| TypeError | operação com tipo incorreto |
| ValueError | valor inválido |
| ZeroDivisionError | divisão por zero |
| FileNotFoundError | arquivo não encontrado |

```python
# exemplo de ValueError

numero = int("abc")
```

# Try / Except

## Bloco try / except

O bloco `try` permite tentar executar um código.

Se ocorrer um erro, o `except` captura esse erro.

```python
try:
    numero = 10 / 0
    print(numero)

except ZeroDivisionError:
    print("Erro: divisão por zero.")
```

## Bloco finally

O `finally` executa **independentemente de ocorrer erro ou não**.

```python
try:
    numero = int("10")
    print(numero)

except ValueError:
    print("Erro de conversão")

finally:
    print("Execução finalizada")
```

## Usando raise

A palavra `raise` permite **gerar um erro manualmente**.

```python
idade = -5

if idade < 0:
    raise ValueError("Idade não pode ser negativa")
```

# Boas práticas

## Prevenção de erros

Sempre que possível devemos **prevenir erros antes que aconteçam**.

```python
numero = input("Digite um número: ")

if numero.isdigit():
    numero = int(numero)
    print(numero)
else:
    print("Valor inválido")
```

## Mensagens claras

Mensagens de erro devem ser **simples e claras** para o usuário.

```python
try:
    numero = int(input("Digite um número: "))
    print(10 / numero)

except ZeroDivisionError:
    print("Não é possível dividir por zero.")

except ValueError:
    print("Digite apenas números.")
```

## Manutenção de código

Um bom tratamento de exceções torna o código:

- mais seguro
- mais fácil de manter
- mais fácil de depurar

```python
# FUNÇÃO DIVIDIR COM ERROS DEPURADOS

def dividir():
    try:
        a = float(input("Digite o número a ser dividido: "))
        b = float(input("Digite o número divisor: "))
        divisao = a / b
        print (f'O resultado da divisão de {a} / {b} é igual a {divisao:.3f}')
        
    except ZeroDivisionError:
        print("Não é possível dividir por zero.")

    except ValueError:
        print("Digite apenas números.")

dividir()
```

```python
# FUNÇÃO CALCULADORA

def calculadora():
    while True:  # loop infinito até o usuário decidir sair
        print("""
        Escolha a operação:
        1 - Soma
        2 - Subtração
        3 - Multiplicação
        4 - Divisão
        0 - Sair
        """)

        try:
            operacao = int(input("Digite a operação: "))

            if operacao == 0:
                print("Calculadora encerrada.")
                break  # sai do loop

            a = float(input("Digite o primeiro número: "))
            b = float(input("Digite o segundo número: "))

            if operacao == 1:
                resultado = a + b
                print(f'O resultado da soma de {a} + {b} é igual a {resultado:.3f}')
            elif operacao == 2:
                resultado = a - b
                print(f'O resultado da subtração de {a} - {b} é igual a {resultado:.3f}')
            elif operacao == 3:
                resultado = a * b
                print(f'O resultado da multiplicação de {a} x {b} é igual a {resultado:.3f}')
            elif operacao == 4:
                resultado = a / b
                print(f'O resultado da divisão de {a} / {b} é igual a {resultado:.3f}')
            else:
                print("Operação inválida.")

        except ZeroDivisionError:
            print("Não é possível dividir por zero.")
        except ValueError:
            print("Digite apenas números.")

calculadora()
```
